```python
# cek outlier
#fungsi cek outlier
import scipy.stats as stats

def chek_outlier(data,columns):
    plt.figure(figsize=(16,4))

    # Histogram
    plt.subplot(1,3,1)
    sns.histplot(data[columns],bins=20, kde=True)
    plt.title('Histogram')

    # QQ-plot
    plt.subplot(1,3,2)
    stats.probplot(data[columns], dist='norm', plot=plt)
    plt.title('QQ-Plot')

    # Boxplot
    plt.subplot(1,3,3)
    sns.boxplot(data[columns])
    plt.title('Box Plot')

    plt.show()
                
    
```

```python
# cek outlier kolom sales

chek_outlier(data,'sales')
```

- terlihat ada outlier dikolom sales
- kita tidak akan menghandling outlier dikolom sales karena, setiap konsumen memiliki daya beli yang berbeda-beda.

```python
# cek outlier dikolom quantity

chek_outlier(data,'quantity')
```

Dikolom quantity tidak terdapat outlier karena nilai quantity sangat masuk akal.

```python
# cek outlier dikolom discount

chek_outlier(data,'discount')
```

dikolom discount tidak terdapat outlier

```python
# cek outlier dikolom profit

chek_outlier(data,'profit')
```

ada outlier dikolom profit, ini sangat mungkin terjadi karena ada banyak faktor yang mempengaruhi. perlu analisis lebih lanjut
